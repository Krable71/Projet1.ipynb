```python
import numpy as np
import matplotlib.pyplot as plt

# 17 catégories
categories =  [
    "Bas-Saint-Laurent", "Saguenay", "Capitale-Nationale",
    "Mauricie", "Estrie", "Montréal",
    "Outaouais", "Abitibi-Témiscamingue",
    "Côte-Nord", "Nord-du-Québec",
    "Gaspésie", "Chaudière-Appalaches",
    "Laval", "Lanaudière",
    "Laurentides", "Montérégie", "Centre-du-Québec" ]

# Exemple de données (17 valeurs pour chaque année)
data_2021 = [8,7,10,7,9,1,5,7,1,0,0,2,2,4,14,4,1]
data_2022 = [2,8,1,4,7,5,15,1,0,0,0,1,0,10,12,7,4]
data_2023 = [1,19,6,8,2,3,6,16,0,3,3,4,0,4,5,9,3]
data_2024 = [1,8,11,5,4,2,6,6,1,1,2,10,1,8,8,5,3]

# Position des catégories sur l'axe x
x = np.arange(len(categories))

# Largeur des barres
width = 0.2

plt.figure(figsize=(12,8))

# Barres pour chaque année
plt.bar(x - 1.5*width, data_2021, width, label="2022")
plt.bar(x - 0.5*width, data_2022, width, label="2023")
plt.bar(x + 0.5*width, data_2023, width, label="2024")
plt.bar(x + 1.5*width, data_2024, width, label="2025")

# Étiquettes
plt.xticks(x, categories, rotation=45, ha="right")
plt.xlabel("")
plt.ylabel("Nombre d'évènement")
plt.title("Nombre d'évènement violent par région administrative du Québec")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pas de temps (axe x)
temps = ["mai", "Juin","Juillet", "Aout", "Septembre"]

# Données pour 4 catégories
categorie1 = [16, 15, 42, 10, 1]
categorie2 = [0, 7, 54, 17, 0]
categorie3 = [13, 28, 29, 17, 5]
categorie4 = [5, 16, 45, 7, 3]

# Création du graphique
plt.plot(temps, categorie1, marker='o', label='2022')
plt.plot(temps, categorie2, marker='o', label='2023')
plt.plot(temps, categorie3, marker='o', label='2024')
plt.plot(temps, categorie4, marker='o', label='2025')

# Personnalisation
plt.xlabel("")
plt.ylabel("Nombre d'évènement ")
plt.title("Évolution mensuelle du nombre d'évènement de tmeps violent ")
plt.legend()

# Affichage
plt.show()
```

```python
import matplotlib.pyplot as plt

years = [2022, 2023, 2024, 2025]
jours_violents = [42, 56, 56, 54]
moyenne_cas = [4.98, 5.3, 6.07, 4.89]

fig, ax1 = plt.subplots(figsize=(6,4))

# Barres avec label
bars = ax1.bar(
    years,
    jours_violents,
    width=0.4,
    label="Nombre de jours de temps violent",
    zorder=1
)

ax1.set_ylabel("Nombre de jours de temps violent", fontsize=11)
ax1.set_xticks(years)

# Ligne avec label
line = ax2 = ax1.twinx()
line_plot = ax2.plot(
    years,
    moyenne_cas,
    marker='o',
    color='orange',
    linewidth=2.5,
    label="Moyenne de cas par jour",
    zorder=3
)

ax2.set_ylabel("Moyenne de cas", fontsize=11)

plt.title("\n\n Évolution des événements de temps violent (2022–2025)", fontsize=12)

# Récupération correcte des éléments
handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    handles1 + handles2,
    labels1 + labels2,
    loc='upper center',
    bbox_to_anchor=(0.5, 1.2),
    ncol=2,
    frameon=False,
    fontsize=10
)

# Style
ax1.spines['top'].set_visible(False)
ax2.spines['top'].set_visible(False)
ax1.grid(alpha=0.3)

plt.subplots_adjust(top=0.78)

plt.savefig("graphique_temps_violent.png", dpi=300, bbox_inches='tight')
plt.show()
```
